# CAPE-Churn — Notebook 01: Exploratory Data Analysis (EDA)

This notebook provides exploratory analysis on the IBM Telco Customer Churn dataset to understand distributions, demographic influences, service plan features, and correlation with customer churn.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv('../backend/data/telco_churn.csv')
print(f"Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns")
df.head()
```

```python
# Target distribution
churn_counts = df['Churn'].value_counts()
print("Churn Distribution:")
print(churn_counts)
print("Churn Percentage:", df['Churn'].value_counts(normalize=True) * 100)
```
